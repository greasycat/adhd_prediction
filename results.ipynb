```python
import pandas as pd

# Read your data
df = pd.read_csv('logs/skip_vote_joint_metrics.csv')

# remove validation columns
df = df.drop(columns=['val_accuracy', 'val_f1_score', 'val_precision', 'val_recall', 'val_roc_auc_score'])

# Define column name mapping
column_mapping = {
    'test_accuracy': 'Test Accuracy',
    'test_f1_score': 'Test F1 Score',
    'test_precision': 'Test Precision',
    'test_recall': 'Test Recall',
    'test_roc_auc_score': 'Test ROC AUC'
}

# Rename columns
df = df.rename(columns=column_mapping)

# Convert all values to percentages with 2 decimal places
df = df.apply(lambda x: (x * 100).round(2))

# Display or save
print(df)

# reset index and rename to epoch
df = df.reset_index()
df = df.rename(columns={'index': 'Epoch'})

# save to latex
df.to_latex(buf='logs/skip_vote_joint_metrics.tex', index=False)
```
