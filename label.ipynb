```python
import pandas as pd
from pathlib import Path
```

```python
phenotypic_files = [x for x in Path("data").glob("*.csv")];
display(phenotypic_files)
```

```python
final_df = pd.DataFrame()
for x in phenotypic_files:
    print(x)
    df = pd.read_csv(x)
    if str(x) == "data/Peking_1_TestRelease_phenotypic.csv":

        # rename ID to "ScanDir ID"
        df = df.rename(columns={"ID": "ScanDir ID"})
    display(df.columns)
    final_df = pd.concat([final_df, df])

final_df.iloc[:, 0] = final_df.iloc[:, 0].apply(lambda x: "sub-" + str(x).zfill(7))
final_df.to_csv("data/all_phenotypic.csv", index=False)
```

```python
# check unique values in the DX column
display(final_df["Site"].unique())
NYU_df = final_df[final_df["Site"] == 5]
```

```python
downloaded_id = pd.read_csv("data/downloaded.txt", header=None)[0].tolist()
```

```python
import seaborn as sns
sns.countplot(x="DX", data=NYU_df[NYU_df["ScanDir ID"].isin(downloaded_id)])
```

```python
NYU_df[NYU_df["ScanDir ID"].isin(downloaded_id)][["ScanDir ID", "DX"]].rename(columns={"ScanDir ID": "id", "DX": "label"}).to_csv("data/NYU_labels.csv", index=False)
```
